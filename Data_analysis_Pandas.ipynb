SPATIAL DATA ANALYSIS SECTION WITH GEOPANDAS

```python
import geopandas as gpd
```

```python
NGA_LGA = gpd.read_file("/content/NGA_adm1.zip", layer='NGA_adm2')
```

```python
NGA_LGA
```

```python
akure = NGA_LGA[NGA_LGA['NAME_2'].isin(['Akure South', 'Akure North'])]
```

```python
#convert df into geodataframe
gdf = gpd.GeoDataFrame(df, geometry=gpd.points_from_xy(df.longitude, df.latitude), crs = 'epsg:4326')
```

```python
gdf
```

```python
pd.set_option('display.max_columns', 100)
```

```python
gdf['lga'].unique()
```

```python
#plot
a_gdf= gdf[gdf['lga'].isin(['Akure South', 'Akure North'])]
```

```python
a_gdf.plot()
```

```python
akure
```

```python
## 1. Count facilities per LGA from your Excel GeoDataFrame
health_akure=a_gdf.groupby('lga').size().reset_index(name='count')
```

```python

#Join akure on a_gdf
dfhealth=akure.merge(health_akure, left_on='NAME_2', right_on='lga',
            how='left')
#
```

```python
dfhealth.plot(color= ['grey', 'beige'], edgecolor = 'black')
```

```python
#overlaying maps

fig, ax = plt.subplots()
akure.plot(ax=ax, color='grey', edgecolor='black')
dfhealth.plot(ax=ax, color='blue', label = 'count', markersize=4)
a_gdf.plot(ax=ax, color='white', markersize=1.2)
```
